```python

class state:
    def __init__(self, as_string, owner):
        items = as_string.split()
        self.state_type = items[0]
        self.state_index = int(items[1])
        self.highest_parent = int(items[2])
        self.parent_count = int(items[3])
        self.lowest_child = int(items[4])
        self.child_count = int(items[5])
        self.other = items[6:]
        self.owner = owner

class node:
    def __init__(self, id, states=[]):
        self.id = id
        self.states = states
        self.children = []
    
    def add_state(self, s:state):
        self.states.append(s)

    def add_child(self, n):
        self.children.append(n)
    
    def print_tree(self, visited:set, indent=0):
        print("\t"*indent, "Node: ", self.id)
        visited.add(self.id)
        idt = indent
        if len(set(self.children)) > 1:
            idt += 1
        for c in set(self.children):
            if(c.id in visited):
                # print("Skipping a loop!")
                if c.id != self.id:
                    print("Found a back-edge! (", self.id, " to ", c.id, ")" )
            else:
                c.print_tree(visited, idt)



```

```python
def parse_file(path):
    root = None
    state_to_parent = {}
    states = []
    mode = "HEADER"
    with open(path) as file:
        for line in file:
            if line.strip() == "//":
                break
            if mode == "HEADER":
                if "[ ROOT    0 ]" in line:
                    mode = "NODES"
                    root = node(id=0)
                    current_node = root
                    continue
            elif mode == "NODES":
                if "[" in line: #Start of a new node
                    toks = line.split()
                    id = toks[2] #3rd token is node number
                    current_node = node(id)
                else: 
                    s = state(line,current_node)
                    states.append(s)
                    current_node.add_state(s)
                    state_to_parent[s.state_index] = current_node
                    # print("State: ", s.state_index, "maps to node", current_node.id)
        #Connect the nodes to their children
        for st in states:
            lc = st.lowest_child
            # print("State ", st.state_index, "has lowest child ", lc)
            for i in range(st.child_count):
                # print("\tHas child at ", lc+i)
                # print("\tThat belongs to ",state_to_parent[lc+i].id)
                st.owner.add_child(state_to_parent[lc+i])
        
        print("Parsed ", len(states), "states")
        return root
```
